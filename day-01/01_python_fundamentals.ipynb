## 1. Python Mental Model & Variables
Unlike JS/TS, Python has no variable declaration keywords like `let`, `const`, or `var`. Variables are simply names bound to objects in memory.

### JavaScript Comparison:
- **JS:** `let x = 10; const name = "Alex";`
- **Python:** `x = 10; NAME = "Alex"` *(Python uses `ALL_CAPS` convention for constants, not runtime enforcement)*

````python
# Variables and assignment
user_name = "Alex Rivera"
login_count = 5
account_balance = 249.50
is_active = True
last_login = None  # None is Python's equivalent of null

print(f"User: {user_name} (Type: {type(user_name).__name__})")
print(f"Logins: {login_count} (Type: {type(login_count).__name__})")
print(f"Balance: ${account_balance} (Type: {type(account_balance).__name__})")
print(f"Active: {is_active} (Type: {type(is_active).__name__})")
print(f"Last Login: {last_login} (Type: {type(last_login).__name__})")
````

### Strong Typing: No Implicit Coercion
JavaScript silently coerces types (`"5" - 2 === 3`, `"5" + 2 === "52"`). 
Python is **strongly typed** and will immediately raise a `TypeError` if you attempt invalid operations across types.

````python
try:
    # In JavaScript: "Total: " + 100 -> "Total: 100"
    # In Python: Raises TypeError
    result = "Total: " + 100
except TypeError as e:
    print(f"Caught expected TypeError: {e}")

# Correct Pythonic approach using f-strings or str():
result = f"Total: {100}"
print("Corrected:", result)
````

## 2. Strings & f-Strings
Python strings are **immutable**. Methods return brand-new string objects.

### Key Operations:
- `.strip()`, `.lower()`, `.upper()`, `.replace()`
- `.split(delimiter)` and `delimiter.join(list)` *(note that `join` is called on the delimiter string)*
- f-strings: `f"{expression}"` with inline format specifiers (e.g. `f"{price:.2f}"`)

````python
raw_title = "   production api service   "

# Cleaning & formatting
clean_title = raw_title.strip().title()
print("Clean Title:", clean_title)

# Splitting and Joining
skills = "Python, TypeScript, Docker, SQL"
skill_list = [s.strip() for s in skills.split(",")]
print("Parsed list:", skill_list)

rejoined = " | ".join(skill_list)
print("Joined with pipe:", rejoined)

# f-string formatting
rate = 85.5
hours = 40
print(f"Invoice: {clean_title} | Total: ${rate * hours:,.2f}")
````

## 3. Numbers, Booleans, and `None`
### Key Differences:
1. **Division:** `/` always produces a `float` (`7 / 2 -> 3.5`). Use `//` for integer floor division (`7 // 2 -> 3`).
2. **Logical operators:** Use `and`, `or`, `not` instead of `&&`, `||`, `!`. 
3. **Identity (`is`) vs Value Equality (`==`):** Always check for `None` using `is None` or `is not None`.

````python
# Arithmetic
print("Float division 7 / 2:", 7 / 2)
print("Floor division 7 // 2:", 7 // 2)
print("Exponentiation 2 ** 8:", 2 ** 8)

# Logical Operators
is_admin = True
is_verified = False

if is_admin and not is_verified:
    print("Admin requires verification check")

# Value vs Identity
config = None
print("Is config None?", config is None)
````

## 4. Lists (Dynamic Arrays)
Lists are ordered, mutable sequences. They correspond directly to JavaScript arrays.

| Operation | Python List | JavaScript Array |
| :--- | :--- | :--- |
| Append to end | `lst.append(x)` | `arr.push(x)` |
| Insert at index | `lst.insert(0, x)` | `arr.unshift(x)` |
| Pop from end | `lst.pop()` | `arr.pop()` |
| Pop from index | `lst.pop(0)` | `arr.shift()` |
| Length | `len(lst)` | `arr.length` |
| Membership | `x in lst` | `arr.includes(x)` |

````python
tasks = ["Setup Docker", "Write unit tests", "Deploy to staging"]

# Mutation
tasks.append("Notify team")
tasks.insert(0, "Code review")
print("Tasks after inserts:", tasks)

completed_task = tasks.pop()
print(f"Popped: {completed_task}")
print("Remaining count:", len(tasks))
print("Is 'Setup Docker' pending?", "Setup Docker" in tasks)
````

## 5. Indexing & Slicing `[start:stop:step]`
Slicing works identically on strings, lists, and tuples:
- `[start:stop]` : extracts items from `start` up to (excluding) `stop`
- Negative indices count from the end (`-1` is last item)
- `[::-1]` cleanly reverses the sequence

````python
endpoints = ["/auth", "/users", "/orders", "/payments", "/metrics"]

print("First item [0]:", endpoints[0])
print("Last item [-1]:", endpoints[-1])
print("Slice [1:4]:", endpoints[1:4])
print("First three [:3]:", endpoints[:3])
print("Last two [-2:]:", endpoints[-2:])
print("Every 2nd item [::2]:", endpoints[::2])
print("Reversed [::-1]:", endpoints[::-1])
````

## 6. Tuples (Immutable Records)
Tuples are fixed, immutable sequences defined with parentheses `(a, b)`. 
They are commonly used for database records, fixed schemas, and multiple return values.

````python
# Tuple definition and unpacking
response = (200, "OK", "application/json")
status_code, message, content_type = response

print(f"Status: {status_code} | Msg: {message} | Type: {content_type}")

# Function returning multiple values
def parse_dimensions(dim_str: str) -> tuple[int, int]:
    w, h = dim_str.split("x")
    return int(w), int(h)

width, height = parse_dimensions("1920x1080")
print(f"Parsed width: {width}px, height: {height}px")
````

## 7. Sets (Uniqueness & Fast Membership)
Sets are unordered collections of unique elements with average **O(1)** membership lookup time.

> **Watchout:** `{}` creates an empty dictionary, not an empty set! Use `set()` for an empty set.

````python
required_roles = {"admin", "developer", "editor"}
user_roles = {"developer", "viewer"}

# Set Algebra
common = required_roles & user_roles      # Intersection
all_roles = required_roles | user_roles   # Union
missing = required_roles - user_roles     # Difference

print("Common roles (&):", common)
print("All roles (|):", all_roles)
print("Missing required roles (-):", missing)

# Fast deduplication
tags = ["frontend", "ui", "backend", "ui", "frontend"]
unique_tags = list(set(tags))
print("Deduplicated tags:", unique_tags)
````

## 8. Dictionaries (Key-Value Hash Maps)
Dictionaries correspond to JavaScript objects / Maps. 

### The `KeyError` Trap & Safe `.get()`:
In JavaScript, `obj.nonExistent` evaluates to `undefined`. 
In Python, accessing a missing key via `dict["nonExistent"]` throws a fatal `KeyError`. Always use `.get(key, default)` when a key might be missing.

````python
task = {
    "id": 1,
    "title": "Implement JWT Authentication",
    "completed": False,
    "priority": "High"
}

# Safe access
priority = task.get("priority", "Normal")
assignee = task.get("assignee", "Unassigned")
print(f"Task: {task['title']} | Priority: {priority} | Assignee: {assignee}")

# Iterating items
print("\n--- Task Breakdown ---")
for k, v in task.items():
    print(f"{k:12} : {v}")
````

## 9. Control Flow & Truthiness
Python evaluates empty containers (`[]`, `{}`, `""`, `set()`, `()`), `0`, `0.0`, `False`, and `None` as **falsy**.

### Pythonic Container Checking:
- **Unpythonic:** `if len(tasks) > 0:`
- **Idiomatic Python:** `if tasks:`

````python
pending_tasks = []

if not pending_tasks:
    print("Queue is clear (pending_tasks is falsy)")

# Ternary expression: [value_if_true] if [condition] else [value_if_false]
status_code = 200
status_label = "Success" if 200 <= status_code < 300 else "Error"
print(f"Status: {status_code} ({status_label})")
````

## 10. Functions & The Mutable Default Argument Trap
Python functions are defined with `def` and support default parameters and keyword arguments.

> **CRITICAL GOTCHA:** Never use a mutable object (such as a list or dictionary) as a default parameter value (`def add(item, lst=[])`). In Python, default arguments are evaluated **once at function definition time**, sharing that same instance across every invocation!

````python
# THE TRAP (Demonstration):
def buggy_append(item, target_list=[]):
    target_list.append(item)
    return target_list

print("Call 1:", buggy_append("A"))
print("Call 2:", buggy_append("B"))  # BUG: Returns ['A', 'B'] instead of ['B']!

# THE PYTHONIC FIX: Default to None
def safe_append(item: str, target_list: list[str] | None = None) -> list[str]:
    if target_list is None:
        target_list = []
    target_list.append(item)
    return target_list

print("\nSafe Call 1:", safe_append("A"))
print("Safe Call 2:", safe_append("B"))  # Correct: Returns ['B']
````

## 11. Pythonic Idioms: Comprehensions, Enumerate & Zip
Comprehensions replace `.map()` and `.filter()` pipelines with declarative, optimized syntax.

````python
numbers = [1, 2, 3, 4, 5, 6, 7, 8]

# List comprehension: [transform for item in collection if condition]
squared_evens = [n ** 2 for n in numbers if n % 2 == 0]
print("Squared evens:", squared_evens)

# Dict comprehension
usernames = ["alice", "bob", "charlie"]
user_map = {u: f"{u}@company.internal" for u in usernames}
print("Generated emails:", user_map)

# enumerate() for index + item
print("\n--- Ranked Items ---")
frameworks = ["FastAPI", "React", "TailwindCSS"]
for rank, framework in enumerate(frameworks, start=1):
    print(f"#{rank}: {framework}")

# zip() for parallel iteration
stars = [72000, 65000, 125000]
print("\n--- Stars Pairing ---")
for name, count in zip(frameworks, stars):
    print(f"{name:10} : {count:,} GitHub stars")
````

## 13. Challenge Solution
Expand the cell below to inspect the reference implementation.

````python
# Reference Solution

def add_task(
    task_list: list[dict],
    title: str,
    priority: str = "Normal",
    tags: list[str] | None = None
) -> dict:
    new_id = len(task_list) + 1
    task = {
        "id": new_id,
        "title": title,
        "completed": False,
        "priority": priority,
        "tags": tags if tags is not None else []
    }
    task_list.append(task)
    return task

# 1. Add a new task
add_task(task_db, "Set up FastAPI routing", priority="High", tags=["api", "backend"])

# 2. Extract urgent tasks (High priority and pending)
urgent_tasks = [t for t in task_db if t["priority"] == "High" and not t["completed"]]

# 3. Calculate statistics
total_count = len(task_db)
completed_count = len([t for t in task_db if t["completed"]])
pending_count = total_count - completed_count

# 4. Formatted output
print("=" * 60)
print(f"TASK SUMMARY: {total_count} Total | {completed_count} Done | {pending_count} Pending")
print("=" * 60)
print("\n--- ACTION REQUIRED (Urgent High Priority Tasks) ---")
for idx, t in enumerate(urgent_tasks, start=1):
    tag_str = ", ".join(t["tags"]) if t["tags"] else "none"
    print(f"[{idx}] #{t['id']} {t['title']:<32} (Tags: {tag_str})")
````
